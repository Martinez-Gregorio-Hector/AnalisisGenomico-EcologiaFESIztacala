### 🌲 La historia

Cada otoño, millones de **mariposas monarca** (*Danaus plexippus*) vuelan más de **4,000 km** desde el sur de Canadá y el norte de Estados Unidos hasta los **bosques de oyamel** (*Abies religiosa*) de la **Reserva de la Biosfera Mariposa Monarca**, entre Michoacán y el Estado de México. Ahí se agrupan por millones en las ramas para pasar el invierno.

Las mariposas que hacen el viaje pertenecen a la llamada **"generación Matusalén"**: viven unos 8 meses, mientras que sus bisabuelas del verano vivieron apenas de 2 a 6 semanas. ¡Ninguna de ellas había estado antes en México y aun así llegan a los mismos cerros!

Como no se pueden contar una por una, cada invierno se mide la **superficie (en hectáreas) que ocupan las colonias** y con eso se estima la población. Esa superficie fue de **18.19 ha** en 1996-1997 y de apenas **0.90 ha** en 2023-2024. Entre las amenazas están la **tala ilegal** del bosque, la pérdida de **algodoncillo** (*Asclepias*, la única planta que comen sus orugas), el uso de herbicidas y el **cambio climático**.
