# Operadores lógicos

### 1. ¿Cuáles son los operadores lógicos?

Los operadores lógicos modifican y unen expresiones evaluadas en contexto booleano para crear condiciones más complejas.

### 2. Operador `not`

```python
num = 5
```

```python
num < 10
```

```python
not num < 10
```

### 3. Operador `or`

```python
num1 = 5
num2 = 10
```

```python
num1 < 4
```

```python
num1 < 4 or num2 > 5
```

```python
(num2 < 4) or (num1 > 5) or (num1 > 3)
```

### 4. Operador `and`

```python
num1 = 5
num2 = 10
```

```python
num1 < 4 and num2 > 5
```

```python
num1 < 10 and num2 > 3
```
